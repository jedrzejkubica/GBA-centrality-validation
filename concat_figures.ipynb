```python
import matplotlib.pyplot
import matplotlib.image
```

```python
FIG_PATH = "figures/CSI2A-v2-260522/evidence3_direct0"
```

```python
paths = [
    f"{FIG_PATH}/MMAF_CDF.png",
    f"{FIG_PATH}/NOA_CDF.png",
    f"{FIG_PATH}/HYPCARD_CDF.png",
    f"{FIG_PATH}/DYSCHROM_CDF.png"
]

labels = ["(a)", "(b)", "(c)", "(d)"]


fig, axes = matplotlib.pyplot.subplots(2, 2, figsize=(12, 9))

for ax, path, label in zip(axes.ravel(), paths, labels):
    img = matplotlib.image.imread(path)
    ax.imshow(img)
    ax.axis("off")
    
    ax.text(
        0.02, 0.95, label,
        transform=ax.transAxes,
        fontsize=10,
        fontweight="normal",
        va="top", ha="left"
    )

matplotlib.pyplot.subplots_adjust(wspace=0, hspace=0)
matplotlib.pyplot.savefig(f"{FIG_PATH}/all_CDF.png", bbox_inches="tight", dpi=700)
matplotlib.pyplot.show()
```
